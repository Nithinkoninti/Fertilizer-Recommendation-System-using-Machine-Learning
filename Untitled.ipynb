# Data Pre-processing

```python
import os
import joblib
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from imblearn.over_sampling import SMOTE

MODEL_DIR = "models"
os.makedirs(MODEL_DIR, exist_ok=True)

def preprocess_data(df, is_train=True, apply_smote=True, random_state=42):
    df = df.copy()

    # -----------------------------
    # Drop unnecessary columns if any (e.g., IDs)
    # -----------------------------
    if 'ID' in df.columns:
        df.drop(columns=['ID'], inplace=True)

    # -----------------------------
    # Encode Categorical Features
    # -----------------------------
    categorical_cols = [
        'Soil_Type', 'Crop_Type', 'Crop_Growth_Stage', 'Season',
        'Irrigation_Type', 'Previous_Crop', 'Region', 'Fertilizer_Used_Last_Season'
    ]

    for col in categorical_cols:
        encoder_path = os.path.join(MODEL_DIR, f"{col}_encoder.pkl")

        if is_train:
            le = LabelEncoder()
            df[col] = le.fit_transform(df[col].astype(str))
            joblib.dump(le, encoder_path)
        else:
            le = joblib.load(encoder_path)
            df[col] = le.transform(df[col].astype(str))

    # -----------------------------
    # Feature Matrix
    # -----------------------------
    X = df.drop(columns=['Recommended_Fertilizer'])

    # -----------------------------
    # Standard Scaling for Numeric Features
    # -----------------------------
    numeric_cols = [
        'Soil_pH', 'Soil_Moisture', 'Organic_Carbon', 'Electrical_Conductivity',
        'Nitrogen_Level', 'Phosphorus_Level', 'Potassium_Level',
        'Temperature', 'Humidity', 'Rainfall', 'Yield_Last_Season'
    ]

    scaler_path = os.path.join(MODEL_DIR, 'standard_scaler.pkl')

    if is_train:
        scaler = StandardScaler()
        X[numeric_cols] = scaler.fit_transform(X[numeric_cols])
        joblib.dump(scaler, scaler_path)
    else:
        scaler = joblib.load(scaler_path)
        X[numeric_cols] = scaler.transform(X[numeric_cols])

    # -----------------------------
    # Target
    # -----------------------------
    y = df['Recommended_Fertilizer']

    # -----------------------------
    # Apply SMOTE for balancing
    # -----------------------------
    if is_train and apply_smote:
        smote = SMOTE(random_state=random_state)
        X_res, y_res = smote.fit_resample(X, y)
        print(f"Applied SMOTE. Original samples: {len(y)}, New samples: {len(y_res)}")
        return X_res, y_res

    return X, y
```

```python
X, y = preprocess_data(df, is_train=True)
```

```python
import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix, roc_curve, auc
)
from sklearn.preprocessing import label_binarize

# Ensure results folder exists
os.makedirs("results", exist_ok=True)

# Global DataFrame to store classification metrics
classification_metrics_df = pd.DataFrame(
    columns=['Algorithm', 'Accuracy', 'Precision', 'Recall', 'F1-Score']
)

def calculate_metrics(task_type, algorithm, y_pred, y_test, y_score=None):
   

    # -----------------------------
    # Load target label encoder (if exists)
    # -----------------------------
    encoder_path = os.path.join("models", "label_encoder.pkl")
    if os.path.exists(encoder_path):
        le_target = joblib.load(encoder_path)
        categories = le_target.classes_
    else:
        categories = np.unique(y_test)

    n_classes = len(categories)

    # -----------------------------
    # Classification Metrics
    # -----------------------------
    acc = accuracy_score(y_test, y_pred) * 100
    prec = precision_score(y_test, y_pred, average='macro', zero_division=0) * 100
    rec = recall_score(y_test, y_pred, average='macro', zero_division=0) * 100
    f1 = f1_score(y_test, y_pred, average='macro', zero_division=0) * 100

    global classification_metrics_df
    classification_metrics_df.loc[len(classification_metrics_df)] = [
        algorithm, acc, prec, rec, f1
    ]

    print(f"\n{algorithm} Metrics")
    print("-" * 40)
    print(f"Accuracy : {acc:.2f}%")
    print(f"Precision: {prec:.2f}%")
    print(f"Recall   : {rec:.2f}%")
    print(f"F1-Score : {f1:.2f}%")

    print("\nClassification Report:")
    print(
        classification_report(
            y_test,
            y_pred,
            labels=range(n_classes),
            target_names=categories,
            zero_division=0
        )
    )

    # -----------------------------
    # Confusion Matrix
    # -----------------------------
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=categories,
        yticklabels=categories
    )
    plt.title(f"{algorithm} - Confusion Matrix")
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.tight_layout()
    plt.savefig(f"results/{algorithm.replace(' ', '_')}_confusion_matrix.png")
    plt.show()

    # -----------------------------
    # ROC Curve
    # -----------------------------
    if y_score is not None:
        plt.figure(figsize=(10, 8))

        # -------- Binary classification --------
        if n_classes == 2:
            fpr, tpr, _ = roc_curve(y_test, y_score[:, 1])
            roc_auc = auc(fpr, tpr)
            plt.plot(fpr, tpr, lw=2, label=f"AUC = {roc_auc:.2f}")

        # -------- Multiclass classification --------
        else:
            y_test_bin = label_binarize(y_test, classes=range(n_classes))

            for i in range(n_classes):
                fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_score[:, i])
                roc_auc = auc(fpr, tpr)
                plt.plot(
                    fpr,
                    tpr,
                    lw=2,
                    label=f"{categories[i]} (AUC = {roc_auc:.2f})"
                )

        plt.plot([0, 1], [0, 1], "k--", lw=1)
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title(f"{algorithm} - ROC Curve")
        plt.legend(loc="lower right")
        plt.grid(True)
        plt.tight_layout()
        plt.savefig(f"results/{algorithm.replace(' ', '_')}_roc_curve.png")
        plt.show()
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
import os
import joblib

def train_lda_classifier(
    X_train,
    y_train,
    X_test,
    y_test,
    solver='svd',
    shrinkage=None
):
    model_path = os.path.join(MODEL_DIR, 'lda.pkl')

    if os.path.exists(model_path):
        print("Loading LDA Classifier...")
        model = joblib.load(model_path)
    else:
        print("Training LDA Classifier...")
        model = LinearDiscriminantAnalysis(
            solver=solver,
            shrinkage=shrinkage
        )
        model.fit(X_train, y_train)
        joblib.dump(model, model_path)
        print(f"Model saved to {model_path}")

    y_pred = model.predict(X_test)

    # LDA supports predict_proba for most solvers
    if hasattr(model, "predict_proba"):
        y_score = model.predict_proba(X_test)
    else:
        y_score = None

    calculate_metrics(
        algorithm="Linear Discriminant Analysis",
        y_pred=y_pred,
        y_test=y_test,
        task_type='classification',
        y_score=y_score
    )

    return model
```

```python
train_lda_classifier(
    X_train,
    y_train,
    X_test,
    y_test,
    solver='svd',
    shrinkage=None
)
```

```python
from sklearn.svm import SVC
import os
import joblib

def train_kernel_svm_classifier(
    X_train,
    y_train,
    X_test,
    y_test,
    kernel='rbf',
    C=1.0,
    gamma='scale'
):
    model_path = os.path.join(MODEL_DIR, 'kernel_svm.pkl')

    if os.path.exists(model_path):
        print("Loading Kernel SVM Classifier...")
        model = joblib.load(model_path)
    else:
        print("Training Kernel SVM Classifier...")
        model = SVC(
            kernel=kernel,
            C=C,
            gamma=gamma,
            probability=True,   # needed for predict_proba
            random_state=42
        )
        model.fit(X_train, y_train)
        joblib.dump(model, model_path)
        print(f"Model saved to {model_path}")

    y_pred = model.predict(X_test)
    y_score = model.predict_proba(X_test)

    calculate_metrics(
        algorithm="Kernel SVM",
        y_pred=y_pred,
        y_test=y_test,
        task_type='classification',
        y_score=y_score
    )

    return model
```

```python
train_kernel_svm_classifier(
    X_train,
    y_train,
    X_test,
    y_test,
    kernel='rbf',
    C=1.0,
    gamma='scale'
)
```

```python
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
import os
import joblib
import numpy as np

def train_qda_classifier(
    X_train,
    y_train,
    X_test,
    y_test,
    reg_param=0.0
):
    model_path = os.path.join(MODEL_DIR, 'qda.pkl')

    # Convert pandas → numpy (safe for QDA)
    X_train = np.asarray(X_train, dtype=np.float64)
    X_test = np.asarray(X_test, dtype=np.float64)

    if os.path.exists(model_path):
        print("Loading QDA Classifier...")
        model = joblib.load(model_path)
    else:
        print("Training QDA Classifier...")
        model = QuadraticDiscriminantAnalysis(
            reg_param=reg_param
        )
        model.fit(X_train, y_train)
        joblib.dump(model, model_path)
        print(f"Model saved to {model_path}")

    y_pred = model.predict(X_test)

    # QDA supports predict_proba
    if hasattr(model, "predict_proba"):
        y_score = model.predict_proba(X_test)
    else:
        y_score = None

    calculate_metrics(
        algorithm="Quadratic Discriminant Analysis",
        y_pred=y_pred,
        y_test=y_test,
        task_type='classification',
        y_score=y_score
    )

    return model
```

```python
train_qda_classifier(
    X_train,
    y_train,
    X_test,
    y_test,
    reg_param=0.0
)
```

```python
from sklearn.ensemble import BaggingClassifier, StackingClassifier
from sklearn.linear_model import LogisticRegression
import os
import joblib

def train_bagging_with_logistic_stacking(
    X_train,
    y_train,
    X_test,
    y_test,
    n_estimators=100,
):
    model_path = os.path.join(
        MODEL_DIR,
        'bagging_logistic_stacking1.pkl'
    )

    if os.path.exists(model_path):
        print("Loading Stacking (Bagging and Logistic) model...")
        model = joblib.load(model_path)
    else:
        print("Training Stacking (Bagging and Logistic) model...")

        # -----------------------------
        # Base Model (Bagging)
        # -----------------------------
        bagging_model = BaggingClassifier(
            n_estimators=n_estimators,
            bootstrap=True,
            random_state=42,
            n_jobs=-1
        )

        # -----------------------------
        # Stacking with Logistic Regression
        # -----------------------------
        model = StackingClassifier(
            estimators=[('bagging', bagging_model)],
            final_estimator=LogisticRegression(
                max_iter=1000,
                random_state=42
            ),
            cv=5,
            passthrough=True,
            n_jobs=-1
        )

        model.fit(X_train, y_train)
        joblib.dump(model, model_path)
        print(f"Model saved to {model_path}")

    # -----------------------------
    # Predictions
    # -----------------------------
    y_pred = model.predict(X_test)
    y_score = model.predict_proba(X_test)

    # -----------------------------
    # Evaluation
    # -----------------------------
    calculate_metrics(
        algorithm="Stacking",
        y_pred=y_pred,
        y_test=y_test,
        task_type='classification',
        y_score=y_score
    )

    return model
```

```python
train_bagging_with_logistic_stacking(
    X_train,
    y_train,
    X_test,
    y_test,
    n_estimators=100, 
)
```

```python
import os
import joblib
import numpy as np

def preprocess_test_data(test_df):
    test_df = test_df.copy()

    # Drop ID if exists
    if 'ID' in test_df.columns:
        test_df.drop(columns=['ID'], inplace=True)

    categorical_cols = [
        'Soil_Type', 'Crop_Type', 'Crop_Growth_Stage', 'Season',
        'Irrigation_Type', 'Previous_Crop', 'Region',
        'Fertilizer_Used_Last_Season'
    ]

    # -----------------------------
    # Load & apply encoders safely
    # -----------------------------
    for col in categorical_cols:
        encoder_path = os.path.join(MODEL_DIR, f"{col}_encoder.pkl")

        if not os.path.exists(encoder_path):
            raise FileNotFoundError(f"Encoder not found for column: {col}")

        if col not in test_df.columns:
            raise ValueError(f"Column '{col}' missing in test data")

        le = joblib.load(encoder_path)

        # Handle unseen categories safely
        test_df[col] = test_df[col].astype(str)

        unseen = set(test_df[col].unique()) - set(le.classes_)
        if unseen:
            print(f"⚠️ Unseen categories in '{col}': {unseen}")
            test_df[col] = test_df[col].apply(
                lambda x: x if x in le.classes_ else le.classes_[0]
            )

        test_df[col] = le.transform(test_df[col])

    # -----------------------------
    # Numeric feature scaling
    # -----------------------------
    numeric_cols = [
        'Soil_pH', 'Soil_Moisture', 'Organic_Carbon', 'Electrical_Conductivity',
        'Nitrogen_Level', 'Phosphorus_Level', 'Potassium_Level',
        'Temperature', 'Humidity', 'Rainfall', 'Yield_Last_Season'
    ]

    scaler_path = os.path.join(MODEL_DIR, 'standard_scaler.pkl')
    if not os.path.exists(scaler_path):
        raise FileNotFoundError("StandardScaler not found")

    scaler = joblib.load(scaler_path)
    test_df[numeric_cols] = scaler.transform(test_df[numeric_cols])

    return test_df
```

```python
def predict_stacking_model(test_df):
    # -----------------------------
    # Load trained stacking model
    # -----------------------------
    model_path = os.path.join(MODEL_DIR, 'bagging_logistic_stacking1.pkl')
    if not os.path.exists(model_path):
        raise FileNotFoundError("Stacking model not found")

    model = joblib.load(model_path)

    # -----------------------------
    # Preprocess test data
    # -----------------------------
    X_test = preprocess_test_data(test_df)

    # -----------------------------
    # Predict
    # -----------------------------
    y_pred = model.predict(X_test)

    # Prediction confidence
    y_proba = model.predict_proba(X_test)
    confidence = np.max(y_proba, axis=1)

    # -----------------------------
    # Attach predictions
    # -----------------------------
    results = test_df.copy()
    results['Predicted_Fertilizer'] = y_pred

    return results
```
